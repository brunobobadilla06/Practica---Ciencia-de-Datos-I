```python
# Importar Numpy y Pandas
import pandas as pd
import numpy as np
```

```python
# Creando una serie desde una lista
s_lista = pd.Series([10, 20, 30, 40, 50])
s_lista
#
```

```python
# Creando una serie con indices propios
s_indices = pd.Series([10, 20, 30, 40, 50], index = ['a', 'b', 'c', 'd', 'e'])
s_indices
```

```python
# creando series desde un diccionario
s_dict = pd.Series({'a': 10, 'b': 20, 'c': 30, 'd': 40, 'e': 50})
s_dict
```

```python
# Creado un data frame desde un diccionario de listas
data = {
    'Nombre': ['Juan', 'Ana', 'Pedro', 'Maria'],
    'Edad': [25, 30, 35, 40],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
}
df = pd.DataFrame(data)
df
```

```python
# Creando un data frame con indices personalizados
df_indice = pd.DataFrame(data, index = ['a', 'b', 'c', 'd'])
df_indice
```
