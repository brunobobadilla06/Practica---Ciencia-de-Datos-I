```python
# Configuracion Inicial
import pandas as pd
# URL del archivo CSV
url = 'https://raw.githubusercontent.com/brunobobadilla06/Practica---Ciencia-de-Datos-I/refs/heads/main/Practica%20Data%20Frame%20N%202.csv'

df_stocks = pd.read_csv(url)
```

```python
# Convertimos la xolumna date en datetime
df_stocks['formatted_date'] = pd.to_datetime(df_stocks['formatted_date'])

# Miramos si se convirtio en datetime
df_stocks.dtypes['formatted_date']
```

```python
# Exploracion basica
# Ver las dimensiones del dataframe
print("Dimesiones del data frame ", df_stocks.shape)
```

```python
# Columnas disponibles
print("Columnas disponibles ", df_stocks.columns.tolist())
```

```python
# Obtener estadisticas basicas de las columnas numericas
print(" Estadisticas Basicas")
df_stocks[['MCD', 'SBUX', 'GOOG', 'AMZN', 'MSFT', 'JPM', 'BAC', 'C', 'MAR', 'HLT', 'RCL', 'V', 'MA', 'PYPL']].describe()
```

Agrupacion Groupby


```python
# Agrupacion de componentes fecha
# Necesitamos extraer componentes de fecha
df_stocks['year'] = df_stocks['formatted_date'].dt.year
df_stocks['month'] = df_stocks['formatted_date'].dt.month
df_stocks['day'] = df_stocks['formatted_date'].dt.day
df_stocks['quarter'] = df_stocks['formatted_date'].dt.quarter
df_stocks
```

```python
ticker_cols = ['MCD', 'SBUX', 'GOOG', 'AMZN', 'MSFT', 'JPM', 'BAC', 'C', 'MAR', 'HLT', 'RCL', 'V', 'MA', 'PYPL']

# Calcular el precio promedio por año para cada ticker (Columnas numericas)
# Agrupamos por año y luego calculamos la media
promedio_anual_tickers = df_stocks.groupby('year')[ticker_cols].mean()
promedio_anual_tickers
```

```python
# Calcular el precio maximo por trimetre por cada ticker
maximo_trimestral_tickers = df_stocks.groupby(['year', 'quarter'])[ticker_cols].max()
maximo_trimestral_tickers
```

```python
# Contar cuantos registros hay en nuestro dataframe por año
conteo_anual = df_stocks.groupby('year').size()
conteo_anual
```

Agregaciones multiples con .agg()

```python
# Calcular el precio promedio y la desviacion estandar anual para GOOG Y MSFT
agreg_seleccion = {
    'GOOG':['mean', 'std'],
    'MSFT':['mean', 'std']
}

promedio_desviacion_anual_ticker = df_stocks.groupby('year').agg(agreg_seleccion)
promedio_desviacion_anual_ticker
```

```python
# Calcular el promedio anual de todos pero el maximo de AMZN

# <Definimos agregaciones para todas las columnas de tickers y una especifica para AMZN
agg_dict = {ticker: 'mean' for ticker in ticker_cols} # Promedio para todos
agg_dict['AMZN'] = ['mean', 'max'] # promedio y maximo para AMZN que reescribe la anterior
```

```python
agg_dict
```

```python
resumen_combinado = df_stocks.groupby('year').agg(agg_dict)
resumen_combinado
```

Aplicando Funciones a datos: apply y map


```python
import pandas as pd
import numpy as np

df = pd.DataFrame({'Ciudad': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Miami'],
                   'Poblacion': [8398748, 3990456, 7894562, 654789, 7896321]})
df
```

```python
# Mapeo usando un diccionario
mapeo_ciudades = {
    'New York': 'Norte',
    'Los Angeles': 'Oeste',
    'Chicago': 'Norte',
    'Houston': 'Oeste',
    'Miami': 'Sur'
}
df['Punto_cardinal'] = df['Ciudad'].map(mapeo_ciudades)
df
```

```python
mapeo_ciudades = {'New York':1, 'Los Angeles':2, 'Chicago':3, 'Houston':4, 'Miami':5}
df['Codigo_Ciudad'] = df['Ciudad'].map(mapeo_ciudades)
df
```

```python
df_numerico = df[['Poblacion', 'Codigo_Ciudad']]

def formato_miles(valor):
  return f'{valor:,}'

df_formateo = df_numerico.map(formato_miles)
df_formateo
```

```python
# apply puede aplicar una fyuncion  lo largo de un eje (filas o columnas)
# axis = 0 aplica acada columna
# axis = 1 aplica a cada fila

def categorizar_poblacion(poblacion):
  if poblacion < 1000000:
    return 'Pequeña'
  elif poblacion < 5000000:
    return 'Mediana'
  else:
    return 'Grande'

df['Tamaño_Poblacion'] = df['Poblacion'].apply(categorizar_poblacion)
df
```

Matplotlib - Fundamentos

```python
# Importar las librerias necesarias
import matplotlib.pyplot as plt
import numpy as np

%matplotlib inline
```

```python
# Geafico de linea simple
x = np.linspace(0, 10, 100) # 100 puntos equidistantes entre 0 y 20
y = np.sin(x) # Calculamos el seno de cada punto de x

# Crear grafico
plt.plot(x, y)

# Etiquetar y poner titulo
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.title('Grafico de linea simple')

# Mostrar el grafico
plt.show()

```

La anatomia de un grafico: Figura y Ejes(Axes)

Figure(Figura): Es el contenedor de mas alto nivel, como la ventana o la pagina completa donde todo se dibuja. Pensemos en la figura como el liezo blanco

Axes(Ejes): Es el area dentro de la figura donde se dibujan los datos(el grafico en si, con sus ejes x e y, etiquetas, etc) Una figura puede contener uno o mas Axes.

Tipos de Graficos mas comunes en Matplotlib OO

```python
# Grafico de dispersion: Relaciona variables numericas
# scatter Plot
x_scatter = np.random.rand(50) * 10 # 50 numeros aleatorios entre 0 y 10
y_scatter = x_scatter + np.random.randn(50) * 2 # y = x + ruido gaussiano

# Crear figura y ejes
fig, ax = plt.subplots()

# Crear grafico de dispersion
ax.scatter(x_scatter, y_scatter)

#Etiquetamos y titulamos
ax.set_xlabel('Eje X')
ax.set_ylabel('Eje Y')
ax.set_title('Grafico de dispersion')

# Mostramos el grafico
plt.show()
```

```python
# Grafico de barras
# Para comparar cantidades en diferentes categorias

# Datos de ejemplo
categorias = ['Manzana', 'Bananas', 'Naranjas', 'Uvas']
cantidades = [25, 40, 30, 35]

# Crear figuras y ejes
fig, ax = plt.subplots()

# Crear grafico de barras
ax.bar(categorias, cantidades)

# Etiquetamos y titulamos
ax.set_xlabel('Categorias')
ax.set_ylabel('Cantidad')
ax.set_title('Grafico de barras')

# Opcional: Rotar etiquetas del eje X si son largas
plt.xticks(rotation=45, ha='left')

# Mostrar el grafico
plt.show()

# Grafico de barras Horizontal
fig, ax = plt.subplots()
ax.barh(categorias, cantidades)

# Etiquetas del grafico de barras horizontales ejes invertidos
ax.set_xlabel('Cantidad')
ax.set_ylabel('Categorias')
ax.set_title('Grafico de barras horizontales')

# Mostrar el grafico
plt.show()
```

Histograma
Muestra la distribucion de una unica variable numerica, agrupando los datos en intervalo

```python
# Datos de ejemplo(distribucion normal)
data = np.random.randn(1000) # 1000 valores aleatorios de una distribucion normal

# Crear figura y ejes
fig, ax = plt.subplots()

# Crear el histograma
ax.hist(data, bins=30, edgecolor='black') # bins controla el numero de barras

# Etiquetamos y titulamos
ax.set_xlabel('Valores')
ax.set_ylabel('Frecuencia')
ax.set_title('Histograma')

#Mostrar grafico
plt.show()
```

Grafico circular
Muestra proporciones de un todo

```python
# datos de ejemplo
etiquetas_pie = ['Manzana', 'Bananas', 'Naranjas', 'Uvas']
cantidades_pie = [25, 40, 30, 35]
colores_pie = ['lightblue', 'lightgreen', 'orange', 'pink']
explode_pie = (0.1, 0, 0, 0) # Explota la primera porción

# Crear figura y ejes
fig, ax = plt.subplots()

# Crear el grafico circular
ax.pie(cantidades_pie, labels=etiquetas_pie, colors=colores_pie, explode=explode_pie, autopct='%1.1f%%', shadow=True, startangle=140)

#Asegurar que el pie sea un circulo
ax.axis('equal')

# Etiquetamos y titulamos
ax.set_title('Frutas y cantidades')

# Mostramos
plt.show()
```

Personalizacion de graficos



Ofrece colores, estilos de linea y marcadores para plot

```python
from matplotlib.lines import lineStyles
x = np.linspace(0, 10, 50)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.sin(x * 2) * 0.5

fig, ax = plt.subplots()

# Graficar personalizaciones
ax.plot(x, y1, color='blue', linestyle='-', marker='', label='Seno') # Linea solida azul
ax.plot(x, y2, color='red', linestyle='--', marker='o', label='Coseno') # Linea discontinua roja con circulos
ax.plot(x, y3, color='#00FF00', linestyle=':', marker='x', label='Seno doble') # Linea punteada verde con cruces

# Añadir leyenda (Usa las etiquetas label definidas en plot)
ax.legend()

# Añadior cuadricula
ax.grid(True, linestyle='--', alpha=0.6) # Cuadricula semitransparente

# Establecer limites de los ejes
ax.set_xlim(0, 10)
ax.set_ylim(-1.5, 1.5)

#etiquetas y titulos
ax.set_xlabel('Eje X')
ax.set_ylabel('Eje Y')
ax.set_title('Grafico Personalizado')

# Mostrar grafico
plt.show()

```

Multiples Graficos (suboplots)

La forma mas comun es plt.subplots()

```python
# Crear datos
x = np.linspace(0, 10, 100)
y_sin = np.sin(x)
y_cos = np.cos(x)
data_hist = np.random.randn(1000)
categorias = ['A', 'B', 'C', 'D']
cantidades = [10, 20, 30, 40]

# Crear figura y subplots 2x2
# figsize controla el tamaño total de la figura en pulgadas
fig, axes = plt.subplots(2, 2, figsize=(10, 8))

# axes ahora es un array de numpy 2x2 que contiene los objetos axes
# accedemos a ellos como axes[fila, columna]

# subplots 1 (Arriba izquierda)
axes[0,0].plot(x, y_sin, color='blue')
axes[0,0].set_title('Seno')
axes[0,0].grid(True)

# subplots 2 (Arriba derecha )
axes[0,1].plot(x, y_cos, color='red')
axes[0,1].set_title('Coseno')
axes[0,1].grid(True)

# subplots 3 (Abajo izquierda)
axes[1, 0].hist(data_hist, bins=20, edgecolor='black', color='green')
axes[1, 0].set_title('Histograma')


# subplots 4 (Abajo derecha)
axes[1, 1].bar(categorias, cantidades, color='orange')
axes[1, 1].set_title('Barras')


# añadir titulo general
fig.suptitle('Multiples graficos en subplots', fontsize=16)

# Mostrar la figura
plt.show()
```
